# Divide and Conquer | Recursive Algorithms

## The main idea

**Divide and conquer** is an algorithm design technique that solves a large problem by working on smaller versions of it:

1. **Divide:** Break the problem into smaller subproblems.
2. **Conquer:** Solve those subproblems, often recursively.
3. **Combine:** Use the smaller results to obtain the final answer (when needed).

Every recursive solution needs a **base case**: a simple situation that can be solved directly and stops the recursion.
